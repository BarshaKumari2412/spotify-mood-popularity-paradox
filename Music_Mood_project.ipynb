```python

import pandas as pd

url = "https://raw.githubusercontent.com/sai-chaitanya-reddy/spotify-tracks-dataset/main/spotify-tracks-dataset-detailed.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Total rows and columns:", df.shape)

display(df.head())
```

```python

print("Column names:")
print(df.columns.tolist())

required_columns = [
    "popularity",
    "valence",
    "energy",
    "danceability"
]

print("\nRequired columns available:")
for col in required_columns:
    print(col, ":", col in df.columns)
```

```python

# Step 3.1: Basic dataset information

print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nFirst 5 rows:")
display(df.head())
```

```python

# Step 3.2: Data quality check

print("Missing values in each column:")
print(df.isnull().sum().sort_values(ascending=False).head(15))

print("\nTotal duplicate rows:", df.duplicated().sum())

required_columns = [
    "popularity",
    "valence",
    "energy",
    "danceability"
]

print("\nRequired columns check:")
for col in required_columns:
    print(f"{col}: {col in df.columns}")
```

```python

# Step 3.3: Check invalid values

for col in ["valence", "energy", "danceability"]:
    if col in df.columns:
        invalid = ~df[col].between(0, 1) & df[col].notna()
        print(f"{col} invalid values:", invalid.sum())

if "popularity" in df.columns:
    invalid_popularity = (
        ~df["popularity"].between(0, 100)
        & df["popularity"].notna()
    )
    print("Popularity invalid values:", invalid_popularity.sum())
```

```python

import pandas as pd

# 1. Check dataset size
print("Original dataset shape:", df.shape)
print("Total rows:", len(df))
print("Total columns:", len(df.columns))

# 2. Check missing values
print("\nMissing values in important columns:")
important_cols = [
    "popularity", "valence", "energy", "danceability"
]

for col in important_cols:
    if col in df.columns:
        print(f"{col}: {df[col].isna().sum()}")

# 3. Check duplicate rows
print("\nExact duplicate rows:", df.duplicated().sum())

# 4. Check data types
print("\nData types:")
print(df[important_cols].dtypes)

# 5. Check value ranges
print("\nSummary statistics:")
display(df[important_cols].describe())
```

```python

# Create a separate copy for analysis
clean_df = df.copy()

# Convert analysis features to numeric
for col in important_cols:
    clean_df[col] = pd.to_numeric(
        clean_df[col], errors="coerce"
    )

# Remove rows missing essential analysis values
clean_df = clean_df.dropna(subset=important_cols)

# Keep valid feature ranges
valid_rows = (
    clean_df["popularity"].between(0, 100)
    & clean_df["valence"].between(0, 1)
    & clean_df["energy"].between(0, 1)
    & clean_df["danceability"].between(0, 1)
)

clean_df = clean_df.loc[valid_rows].copy()

# Remove exact duplicate records only
clean_df = clean_df.drop_duplicates().reset_index(drop=True)

print("Original rows:", len(df))
print("Cleaned rows:", len(clean_df))
print("Rows removed:", len(df) - len(clean_df))
print("Remaining missing values:")
print(clean_df[important_cols].isna().sum())

display(clean_df.head())
```

```python

# Step 4.1: Basic statistical analysis

features = ["popularity", "valence", "energy", "danceability"]

print("Dataset shape:", clean_df.shape)

print("\nStatistical summary:")
display(clean_df[features].describe().round(3))

print("\nCorrelation with popularity:")
correlations = (
    clean_df[features]
    .corr(numeric_only=True)["popularity"]
    .drop("popularity")
    .sort_values(ascending=False)
)

display(correlations.to_frame("Correlation with Popularity"))
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(1, 3, figsize=(17, 5))

sns.regplot(
    data=clean_df, x="valence", y="popularity",
    scatter_kws={"alpha": 0.15, "s": 10},
    line_kws={"color": "red"}, ax=axes[0]
)
axes[0].set_title("Valence vs Popularity")

sns.regplot(
    data=clean_df, x="energy", y="popularity",
    scatter_kws={"alpha": 0.15, "s": 10},
    line_kws={"color": "red"}, ax=axes[1]
)
axes[1].set_title("Energy vs Popularity")

sns.regplot(
    data=clean_df, x="danceability", y="popularity",
    scatter_kws={"alpha": 0.15, "s": 10},
    line_kws={"color": "red"}, ax=axes[2]
)
axes[2].set_title("Danceability vs Popularity")

plt.tight_layout()
plt.show()
```

```python

# Define four mood combinations

conditions = [
    (clean_df["valence"] >= 0.5) & (clean_df["energy"] >= 0.5),
    (clean_df["valence"] >= 0.5) & (clean_df["energy"] < 0.5),
    (clean_df["valence"] < 0.5) & (clean_df["energy"] >= 0.5),
    (clean_df["valence"] < 0.5) & (clean_df["energy"] < 0.5)
]

labels = [
    "Positive & High Energy",
    "Positive & Low Energy",
    "Low Valence & High Energy",
    "Low Valence & Low Energy"
]

clean_df["mood_group"] = "Other"

import numpy as np
clean_df["mood_group"] = np.select(
    conditions, labels, default="Other"
)

mood_summary = (
    clean_df.groupby("mood_group", observed=True)
    .agg(
        song_count=("popularity", "size"),
        avg_popularity=("popularity", "mean"),
        median_popularity=("popularity", "median")
    )
    .sort_values("avg_popularity", ascending=False)
)

display(mood_summary.round(2))

mood_summary["avg_popularity"].plot(
    kind="bar", figsize=(10, 5)
)
plt.title("Average Popularity by Music Mood")
plt.xlabel("Mood Group")
plt.ylabel("Average Popularity")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```

```python

mood_summary.to_csv("mood_popularity_summary.csv")

print("Mood analysis summary saved!")
```

```python

# STEP 6: Compare combinations of mood and audio features

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

analysis_df = clean_df.copy()

# Divide songs into high and low groups
analysis_df["valence_group"] = analysis_df["valence"].apply(
    lambda x: "High valence" if x >= 0.5 else "Low valence"
)

analysis_df["energy_group"] = analysis_df["energy"].apply(
    lambda x: "High energy" if x >= 0.5 else "Low energy"
)

analysis_df["danceability_group"] = analysis_df["danceability"].apply(
    lambda x: "High danceability" if x >= 0.5 else "Low danceability"
)

# Calculate popularity for each combination
combo_summary = (
    analysis_df.groupby(
        ["valence_group", "energy_group", "danceability_group"],
        observed=True
    )
    .agg(
        song_count=("popularity", "size"),
        avg_popularity=("popularity", "mean"),
        median_popularity=("popularity", "median")
    )
    .reset_index()
    .sort_values("avg_popularity", ascending=False)
)

print("Popularity ranking for all feature combinations:")
display(combo_summary.round(2))

print("Highest average popularity combination:")
display(combo_summary.head(1).round(2))

print("Lowest average popularity combination:")
display(combo_summary.tail(1).round(2))

# Visualize average popularity by valence and energy
heatmap_data = analysis_df.pivot_table(
    index="energy_group",
    columns="valence_group",
    values="popularity",
    aggfunc="mean"
)

plt.figure(figsize=(8, 5))
sns.heatmap(heatmap_data, annot=True, fmt=".2f", cmap="YlGnBu")
plt.title("Average Popularity: Energy vs Valence")
plt.xlabel("Musical Valence")
plt.ylabel("Energy Level")
plt.tight_layout()
plt.show()

# Save results for GitHub
combo_summary.to_csv("mood_energy_danceability_analysis.csv", index=False)
heatmap_data.to_csv("energy_valence_heatmap_data.csv")

print("Step 6 completed. Analysis files saved.")
```
